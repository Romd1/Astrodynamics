```python
import numpy as np
D=10 # m
delta=1/3600 # deg
d=np.tan(np.radians(delta)/2)*D*2
print(D, delta, d)

b = -20;
x = 8;
y = np.sin(np.radians(b * x));

x1 = np.degrees(np.arcsin(y)) / b # = 1

x2 = (np.degrees(np.arcsin(y)) - 180) / b # = 8

x3 = x1 - 360 / b # = 19
print(x1, x2, x3)
print(np.sin(np.radians(b * x1)) == np.sin(np.radians(b * x2)) == np.sin(np.radians(b * x3)))
b * (x1 + x2) == -180
np.mod(b * (x1 - x3), 360) == 0
```
